## Введение
Эта статья - первая статья из цикла о сферах и шарах в разных пространствах. В ней рассмотрим сферы на некоторых конечномерных пространствах. Вперёд!

**Окружность** - это геометрическое место точек плоскости, равноудалённых от одной фиксированной точки, называемой **центром**.

(**todo** картинка окружности)

А **сфера** - это аналогичное множество точек, но не на плоскости, а в пространстве. Например, в обычном трёхмерном пространстве $\mathbb{R}^3$. Или в любом другом пространстве, где мы можем измерить расстояние от точки до точки. То есть, в любом **метрическом пространстве**

В общем случае такое множество точек пространства тоже называется **сфера**.

В зависимости от того, как именно измерять расстояние, окружности и сферы будут принимать разные геометрические формы. В этой статье мы их рассмотрим для некоторых конечномерных случаев. А именно, для  $\mathbb{R}^1, \mathbb{R}^2, \mathbb{R}^3$ и $\mathbb{R}^n$

 (формальное определение метрического пространства чуть ниже).

## Как измеряем расстояние ?
**todo** логику нормы и лин. пространства и связи с метрикой - унести вниз в подраздел о норме 

todo - список нумерованных плохо отображается в деплое
**Опр.1.** **Нормой** называется неотрицательная вещественная функция $\|\cdot\|$ на линейном пространстве $L$ над полем $\mathbb{K}$ (где $\mathbb{K} = \mathbb{R}$ или $\mathbb{K} = \mathbb{C}$), удовлетворяющая следующим условиям:
1. из равенства $\|x\| = 0$ следует, что $x = \vartheta$;
2.  для любых $x, y \in L$ верно $\|x + y\| \leqslant \|x\| + \|y\|$; **(неравенство треугольника)**
3.  для любых $x \in L$, $\lambda \in \mathbb{K}$ верно $\|\lambda x\| = |\lambda| \cdot \|x\|$. **(однородность)**

**Опр.2.** Линейное пространство, снабжённое нормой, называется **нормированным пространством**:
$$
N = (L, \|\cdot\|).
$$

Простыми интуитивными словами, норма выдаёт длину вектора, размер вектора. То есть, это функция, которая каждому вектору из линейного(векторного) пространства сопоставляет неотрицательное число.

Если мы зададим на $L$ функцию, удовлетворяющую определению нормы, то она и будет называться нормой.  На одном и том же линейном пространстве так можно задать разные нормы - тогда одному и тому же вектору они могут сопоставить разные значения его, как бы, длины.

**todo** **прим.2** Говоря о линейном пространстве, удобно представлять обычную плоскость $\mathbb{R}^2$. Где каждая точка - это как бы вектор из нуля до этой точки. Это визуально удобно. Каждой точке в $\mathbb{R}^2$ соответствует единственный вектор в $\mathbb{R}^2$, и наоборот. Почему? Оставим обоснование как упражнение.

**Опр.3.** Множество $Y$ называется **метрическим пространством**, если на нём задана вещественная функция $d: Y \otimes Y \to \mathbb{R}^1_+$ такая, что выполнены следующие свойства:
1. $d(x,y) = 0$ тогда и только тогда, когда $x = y$;
2. $d(x,y) = d(y,x)$ для всех $x, y \in Y$; **(симметричность)**
3. $d(x,y) \leqslant d(x,z) + d(z,y)$ для всех $x, y, z \in Y$.**(неравенство треугольника)**

и эта функция $d$ называется **метрикой**

Простыми словами, метрика это функция, которая принимает две точки, и показывает расстояние между ними. Или, задаёт расстояние между ними.

Например, расстояния в нашей бытовой жизни измеряются евклидовой метрикой ($l_2-метрика$). Математически она выражается вот так:
$$
d(x,y) = \sqrt{\sum_{i=1}^n (x_i - y_i)^2}
$$

Мы привели функции для нормы и метрики, почему бы не обойтись лишь одной из этих функций ? Вопрос хороший. Можно и обойтись (todo проверить). 

Чтобы измерять расстояния, нужна только метрика. Но в обычных линейных пространствах метрику удобно задавать по норме. 

**прим.** Метрическое пространство - более общая абстракция, чем линейное пространство. Метрическое пространство - это просто произвольное множество с заданной на нём метрикой. Это значит, нет требований даже к тому, чтобы можно было даже складывать элементы этого множества или умножать их на число. 

А нормированное пространство - это линейное пространство (совсем не "просто множество", а множество векторов, удовлетворяющих аксиомам лин. пространства) с заданной нормой. А в аксиомах линейного пространства - там много требований. И складывать нужно уметь вектора(элементы линейного пространства), и умножать на число.

При этом, норму можно задать только на линейном пространстве. На произвольном множестве не получилось бы её задать, т.к. аксиомы нормы используют операции сложения векторов и умножения на скаляр - а они определены только в линейном пространстве.

Каждая норма порождает метрику вот так: $d(x,y) = \|x-y\|$, но не каждая метрика порождена нормой. Даже в линейных пространствах можно найти такие метрики, которые не могут быть порождены никакой нормой (примеры, чисто названия таких метрик).


Для нашей же задачи, мы можем задать просто норму на линейном пространстве. А метрику породить уже этой нормой. 

Так, в евклидовом n-мерном пространстве стандартная норма
$$
\|x\|_2 = \sqrt{\sum_{i=1}^n x_i^2}
$$
 порождает евклидову метрику 
$$
d(x,y) = \|x - y\|_2 = \sqrt{\sum_{i=1}^n (x_i - y_i)^2}
$$

## Сферы и шары. Окружности и круги.
Перейдём к сферам. И окружностям. И шарам. И кругам.
**Шар** - это множество точек, расстояние от центра (заранее заданной точки) до которых меньше либо равно $r$(радиусу). **Сфера** - это его граница. То есть, только те точки, расстояние от которых до центра строго равно $r$.

В математике используются термины сфера и шар для таких множеств в любых пространствах.

Но для двухмерного случая есть спец.названия - окружность и круг. Они используется в одних случаях исключительно для сферы/шара в евклидовой метрике на плоскости. В других случаях для любой метрики на плоскости.  В любом случае, эти термины только на плоскости.

В этой статье здесь и далее под словом "окружность" понимается сфера $S(x_0, r)$ в двумерном пространстве $(\mathbb{R}^2, \|\cdot\|_p)$ с произвольной нормой $p$, форма которой в общем случае не является кругом в бытовом смысле.  Кругом будем называть окружность + её внутренность.

Также, далее в этой статье, для простоты будем рассматривать только единичные шары, окружности, сферы и круги, т.е. везде считаем $r=1$.

Строго математически, сфера и шар в метрическом пространстве определяются вот так:
Пусть $(Y, d)$ — метрическое пространство, $x_0 \in Y$, $r > 0$. Тогда:

**Открытый шар:**
$$
B(x_0, r) = \{\, x \in Y : d(x, x_0) < r \,\}
$$

**Замкнутый шар:**
$$
\overline{B}(x_0, r) = \{\, x \in Y : d(x, x_0) \leqslant r \,\}
$$

**Сфера:**
$$
S(x_0, r) = \{\, x \in Y : d(x, x_0) = r \,\}
$$

Здесь $x_0 \in Y$ - центр шара, $r > 0$ - его радиус.
